## Stratified metrics

From site `reports/eplus/zone_validation/zone_validation_stratified.json`.

```python
strat = json.loads((site/'reports/eplus/zone_validation/zone_validation_stratified.json').read_text())
rows = []
for z, blocks in (strat.get('stratified_zone_metrics') or {}).items():
    for name, met in blocks.items():
        if met:
            rows.append({'zone': z, 'slice': name, **met})
pd.DataFrame(rows).sort_values(['zone','slice'])
```
